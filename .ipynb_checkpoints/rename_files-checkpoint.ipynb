```python
from pathlib import Path
import shutil

# ============================================================
# User settings
# ============================================================

source_dir = Path("data/2026-07-30")
output_dir = Path("data/new")

output_dir.mkdir(exist_ok=True)

# ============================================================
# Helpers
# ============================================================

def fmt_voltage(v):
    """
    Convert a voltage to the filename format.

    Example:
        -1.07  -> Xn1d070
         0.176 -> Xp0d176
        -0.35  -> Xn0d350
    """

    sign = "p" if v >= 0 else "n"
    v = abs(v)

    return f"{sign}{v:.3f}".replace(".", "d")


# ============================================================
# Voltage lists from July 30 notebook
# ============================================================

# -----------------------
# B-series
# -----------------------

B_init = (-1.07, 0.176)

B_x_y_values = [
    0.71,
    0.65,
    0.60,
    0.56,
    0.50,
    0.45,
    0.39,
    0.35,
    0.30,
    0.24,
    0.20,
    0.15,
    0.10,
    0.05,
    0.00,
    -0.06,
    -0.10,
    -0.15,
    -0.20,
    -0.25,
    -0.30,
    -0.36,
]

B_y_x_values = [
    -0.40,
    -0.45,
    -0.51,
    -0.55,
    -0.60,
    -0.64,
    -0.70,
    -0.75,
    -0.80,
    -0.85,
    -0.90,
    -0.95,
    -0.99,
    -1.05,
    -1.11,
    -1.15,
    -1.20,
    -1.25,
    -1.30,
    -1.35,
    -1.40,
    -1.45,
    -1.50,
    -1.54,
    -1.60,
    -1.65,
    -1.70,
    -1.74,
    -1.80,
    -1.85,
    -1.90,
    -1.95,
    -2.00,
]

# -----------------------
# T-series
# -----------------------

T_init = (-1.05, 0.20)

T_y_x_values = [
    -1.80,
    -1.75,
    -1.70,
    -1.65,
    -1.60,
    -1.55,
    -1.50,
    -1.45,
    -1.40,
    -1.35,
    -1.30,
    -1.24,
    -1.21,
    -1.15,
    -1.10,
    -1.05,
    -1.00,
    -0.95,
    -0.90,
    -0.85,
    -0.80,
    -0.75,
    -0.70,
    -0.65,
    -0.60,
    -0.55,
    -0.50,
    -0.45,
    -0.40,
    -0.35,
    -0.30,
]

T_x_y_values = [
    0.75,
    0.70,
    0.65,
    0.60,
    0.55,
    0.50,
    0.45,
    0.40,
    0.35,
    0.30,
    0.25,
    0.20,
    0.15,
    0.10,
    0.05,
    0.00,
    -0.05,
    -0.10,
    -0.15,
    -0.20,
    -0.25,
    -0.30,
    -0.35,
    -0.40,
    -0.45,
    -0.50,
    -0.55,
    -0.60,
]

# ============================================================
# Rename helper
# ============================================================

def rename_series(files, x_values, y_values, prefix):

    assert len(files) == len(x_values) == len(y_values)

    for file, x, y in zip(files, x_values, y_values):

        timestamp = file.stem.split("_")[-2] + "_" + file.stem.split("_")[-1]

        new_name = (
            f"{prefix}"
            f"_X{fmt_voltage(x)}"
            f"_Y{fmt_voltage(y)}"
            f"_{timestamp}.tif"
        )

        shutil.copy2(file, output_dir / new_name)


# ============================================================
# Initial images
# ============================================================

for name, (x, y) in [
    ("B_series_init", B_init),
    ("T_series_init", T_init),
]:

    file = next(source_dir.glob(f"{name}_*.tif"))

    timestamp = file.stem.split("_")[-2] + "_" + file.stem.split("_")[-1]

    new_name = (
        f"{name}"
        f"_X{fmt_voltage(x)}"
        f"_Y{fmt_voltage(y)}"
        f"_{timestamp}.tif"
    )

    shutil.copy2(file, output_dir / new_name)

# ============================================================
# B x-series
# ============================================================

files = sorted(source_dir.glob("B_series_x_*.tif"))

rename_series(
    files,
    [-1.07] * len(files),
    B_x_y_values,
    "B_series",
)

# ============================================================
# B y-series
# ============================================================

files = sorted(source_dir.glob("B_series_y_*.tif"))

rename_series(
    files,
    B_y_x_values,
    [0.20] * len(files),
    "B_series",
)

# ============================================================
# T y-series
# ============================================================

files = sorted(source_dir.glob("T_series_y_*.tif"))

rename_series(
    files,
    T_y_x_values,
    [0.20] * len(files),
    "T_series",
)

# ============================================================
# T x-series
# ============================================================

files = sorted(source_dir.glob("T_series_x_*.tif"))

rename_series(
    files,
    [-1.10] * len(files),
    T_x_y_values,
    "T_series",
)

print("Done!")
```
