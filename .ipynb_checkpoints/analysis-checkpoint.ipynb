## 1. Load Camera Images
All camera images are stored in a single experiment directory. The filenames contain the series (B or T), the scan direction (x or y), the x- and y-voltages, and the acquisition timestamp. Images are automatically parsed by `load_series()`.

Each image filename in the directory must follow the naming convention:

`<Series>_series_<Measurement>_X<Voltage>_Y<Voltage>_<Timestamp>.tif`

where
- `B_series` or `T_series` = identifies the image series (`B` or `T`)
   - Do not include for `init` images.
- `init`, `x` or `y` = identifies the measurement type:
   - `init` = initial reference image
   - `x` = x-scan (x fixed, y varied)
   - `y` = y-scan (x varied, y fixed)
- `Xn1d070` = x-axis voltage. The first letter indicates the axis (`X`), `n` indicates a negative voltage (`p` indicates positive), and `d` represents the decimal point. For example:
   - `Xn1d070` $\implies$ **-1.07 V**
   - `Xp0d250` $\implies$ **+0.250 V**
- `Yp0d176` = y-axis voltage using the same convention:
   - `Xn1d070` $\implies$ **+0.176 V**
   - `Yn0d050` $\implies$ **−0.050 V**
- `_2026-07-30T13-46-23.947.tif` = acquisition timestamp in the format `YYYY-MM-DDTHH-MM-SS.sss`.

For example,

```text
B_series_x_Xn1d070_Yp0d176_2026-07-30T13-46-23.947.tif

## 2. Pupil Calibration
The pupil centers and aperture radius only need to be determined once for each camera configuration. These values are then reused for all scane acquired with that configuration.

### 2.1 B-Series Calibration
Choose approximate centers for the left and right pupils, and adjust the aperture radius until it fully contains the illuminated pupil without including excessive background.

Definitions:
* `B_left_center` = (x, y) coordinates of the center of the left pupil
* `B_right_center` = (x, y) coordinates of the center of the right pupil
* `B_radius` =  radius (in pixels) of the circular photometric aperture, assumed to be the same for both pupils

````python
# Change these if needed!
B_left_center = (588, 696)
B_right_center = (1044, 715)
B_radius = 188

print("B series:")
# Plot to check guesses here:
int.plot_crops(
    B_init["image"],
    B_left_center,
    B_right_center,
    B_radius,
)
````

## 2.2 T-Series Calibration

````python
# Change these if needed!
T_left_center = (594, 294)
T_right_center = (1047, 300)
T_radius = 188

print("T series:")
# Plot to check guesses here:
int.plot_crops(
    T_init["image"],
    T_left_center,
    T_right_center,
    T_radius,
)
````

## 3 Measure Pupil Intensities & Plot Intensity Scans
For each image, the integrated intensity inside both circular apertures is computed using aperture photometry. The scan voltage and measured pupil intensities are returned for plotting.

````python
voltage, B_x_left_flux, B_x_right_flux = int.measure_series(
    B_x,
    B_left_center,
    B_right_center,
    B_radius,
)

int.plot_intensity(
    B_x,
    voltage,
    B_x_left_flux,
    B_x_right_flux,
)
````

````python
voltage, B_y_left_flux, B_y_right_flux = int.measure_series(
    B_y,
    B_left_center,
    B_right_center,
    B_radius,
)

int.plot_intensity(
    B_y,
    voltage,
    B_y_left_flux,
    B_y_right_flux,
)
````

````python
voltage, T_x_left_flux, T_x_right_flux = int.measure_series(
    T_x,
    T_left_center,
    T_right_center,
    T_radius,
)

int.plot_intensity(
    T_x,
    voltage,
    T_x_left_flux,
    T_x_right_flux,
)
````

````python
voltage, T_y_left_flux, T_y_right_flux = int.measure_series(
    T_y,
    T_left_center,
    T_right_center,
    T_radius,
)

int.plot_intensity(
    T_y,
    voltage,
    T_y_left_flux,
    T_y_right_flux,
)
````
